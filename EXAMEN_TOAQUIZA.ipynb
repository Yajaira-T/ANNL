```python
import numpy as np

# Definir la función que queremos integrar
def funcion(x):
    return np.abs(x) * np.exp(x)

# a) Método de sumas de Riemann superior e inferior

def sumas_riemann_superior_inferior(a, b, n):
    # Dividimos el intervalo en n subintervalos
    h = (b - a) / n
    x = np.linspace(a, b, n + 1)  # puntos de integración
    
    # Sumas inferiores y superiores
    suma_inferior = 0
    suma_superior = 0
    for i in range(n):
        suma_inferior += funcion(x[i])
        suma_superior += funcion(x[i + 1])
    
    suma_inferior *= h
    suma_superior *= h
    
    return suma_inferior, suma_superior

# Llamada para obtener los valores de la suma inferior y superior
a = -1
b = 1
n = 100  # número de subintervalos
suma_inferior, suma_superior = sumas_riemann_superior_inferior(a, b, n)

print(f"Suma inferior: {suma_inferior}")
print(f"Suma superior: {suma_superior}")
```

La elección de las sumas de Riemann **superiores** o **inferiores** es **independiente** al resultado de la aproximación de la integral en este caso, ya que ambas aproximaciones, al utilizar particiones del intervalo y sumas, convergen hacia el mismo valor a medida que el número de subintervalos aumenta. Este comportamiento se explica debido a varias razones fundamentales:

### 1. Propiedades de la función $$ f(x) = |x| e^x $$
- **Monotonía**: La función $$ f(x) = |x| e^x $$ es continua y positiva en el intervalo $$ [-1, 1] $$ . Aunque tiene un cambio de signo en $$ x = 0 $$, el valor absoluto garantiza que la función es no negativa en todo el intervalo.
- **Comportamiento suave**: La función no presenta saltos, discontinuidades o picos abruptos, lo que hace que las aproximaciones con sumas de Riemann converjan al valor real de la integral.

### 2. Teoría de las sumas de Riemann
Las sumas de Riemann **superiores** e **inferiores** se construyen utilizando aproximaciones de la integral mediante sumas de áreas de rectángulos. En cada caso:
- La **suma inferior** toma el valor mínimo de la función en cada subintervalo.
- La **suma superior** toma el valor máximo de la función en cada subintervalo.

Dado que la función es continua y no tiene irregularidades, la suma superior y la suma inferior se aproximan cada vez más al valor real de la integral a medida que aumenta el número de subintervalos $$ n $$ . Esta convergencia es garantizada por el **teorema de las sumas de Riemann**, que establece que, para funciones continuas en un intervalo cerrado, la diferencia entre las sumas superior e inferior se reduce a medida que el número de subintervalos aumenta, lo que lleva a una aproximación más precisa de la integral.

### 3. Convergencia hacia el valor de la integral
La **diferencia** entre la suma superior y la inferior depende de la **anchura** de los subintervalos $$ (h) $$ y la variabilidad de la función dentro de cada subintervalo. Para una función continua y razonablemente suave como $$ f(x) = |x| e^x $$, estas diferencias disminuyen conforme se aumenta $$ n $$, por lo que ambas sumas convergen al mismo valor exacto de la integral cuando $$ n \to \infty $$.

En términos matemáticos, para una función continua, el **valor exacto** de la integral es igual al límite de las sumas de Riemann, tanto para la suma inferior como para la superior:

$$
\int_{a}^{b} f(x) \, dx = \lim_{n \to \infty} \sum_{i=1}^{n} f(x_i) \Delta x = \lim_{n \to \infty} \sum_{i=1}^{n} f(x_i^*) \Delta x
$$

donde $$ x_i $$ y $$ x_i^* $$ son los puntos de evaluación de las sumas inferior y superior, respectivamente.

### 4. Conclusión
Dado que la diferencia entre la suma inferior y la suma superior se hace arbitrariamente pequeña cuando $$ n $$ aumenta, la elección de usar una u otra es en última instancia **irrelevante** para obtener una aproximación precisa de la integral. Ambas sumas proporcionan una estimación que converge al mismo valor exacto de la integral.

En resumen, **la elección entre sumas superiores o inferiores no afecta el valor de la integral aproximada**, ya que ambas tienden al mismo valor a medida que aumentamos el número de subintervalos.

```python
import numpy as np

# Definir la función que queremos integrar
def funcion(x):
    return np.abs(x) * np.exp(x)

# b) Métodos de punto medio, trapezoide y Cavalieri-Simpson

# Método de punto medio
def metodo_punto_medio(a, b, n):
    h = (b - a) / n
    x = np.linspace(a + h/2, b - h/2, n)  # puntos de evaluación
    return h * np.sum(funcion(x))

# Método trapezoide
def metodo_trapezoide(a, b, n):
    h = (b - a) / n
    x = np.linspace(a, b, n + 1)
    return (h / 2) * (funcion(x[0]) + 2 * np.sum(funcion(x[1:-1])) + funcion(x[-1]))

# Método de Cavalieri-Simpson
def metodo_cavalieri_simpson(a, b, n):
    h = (b - a) / (2 * n)
    x = np.linspace(a, b, 2 * n + 1)
    return (h / 3) * (funcion(x[0]) + 4 * np.sum(funcion(x[1::2])) + 2 * np.sum(funcion(x[2:-2:2])) + funcion(x[-1]))


n = 100  # número de subintervalos
integral_punto_medio = metodo_punto_medio(a, b, n)
integral_trapezoide = metodo_trapezoide(a, b, n)
integral_simpson = metodo_cavalieri_simpson(a, b, n)

print(f"Integración por punto medio: {integral_punto_medio}")
print(f"Integración por trapezoide: {integral_trapezoide}")
print(f"Integración por Simpson: {integral_simpson}")
```

### Literal c

```python
import numpy as np
import matplotlib.pyplot as plt

# Definir la función que queremos integrar
def f(x):
    return np.abs(x) * np.exp(x)

# Métodos de integración 
def sumas_riemann_superior_inferior(a, b, n):
    h = (b - a) / n
    x = np.linspace(a, b, n + 1)
    suma_inferior = 0
    suma_superior = 0
    for i in range(n):
        suma_inferior += f(x[i])
        suma_superior += f(x[i + 1])
    suma_inferior *= h
    suma_superior *= h
    return (suma_inferior + suma_superior) / 2  # Promedio para Riemann

def metodo_punto_medio(a, b, n):
    h = (b - a) / n
    x = np.linspace(a + h/2, b - h/2, n)
    return h * np.sum(f(x))

def metodo_trapezoide(a, b, n):
    h = (b - a) / n
    x = np.linspace(a, b, n + 1)
    return (h / 2) * (f(x[0]) + 2 * np.sum(f(x[1:-1])) + f(x[-1]))

def metodo_cavalieri_simpson(a, b, n):
    # n debe ser par para Simpson
    if n % 2 != 0:
        n += 1
    h = (b - a) / n
    x = np.linspace(a, b, n + 1)
    
    weights = np.ones(n + 1)
    weights[1:-1:2] = 4  # índices impares
    weights[2:-2:2] = 2  # índices pares (excepto primero y último)
    return (h / 3) * np.sum(weights * f(x))

# Valor exacto de la integral
def valor_exacto():
    return 2 - 2/np.e

# c) Estudio de convergencia para precisión de 7 cifras decimales
def estudio_convergencia():
    a, b = -1, 1
    exacto = valor_exacto()
    print(f"Valor exacto de la integral: {exacto:.10f}")
    
    # Precisión deseada: 7 cifras decimales
    precision = 1e-7
    
    # Listas para almacenar resultados
    metodos = ['Riemann', 'Punto Medio', 'Trapezoide', 'Simpson']
    funciones = [sumas_riemann_superior_inferior, metodo_punto_medio, 
                 metodo_trapezoide, metodo_cavalieri_simpson]
    
    # Diccionario para almacenar resultados
    resultados = {metodo: {'nodos': [], 'valores': [], 'errores': []} 
                  for metodo in metodos}
    
    # Número máximo de nodos a probar
    max_nodos = 500
    
    # Para cada número de subintervalos
    for n in range(1, max_nodos + 1):
        # Nodos = subintervalos + 1
        nodos = n + 1
        
        # Para Simpson, necesitamos n par
        for metodo, funcion in zip(metodos, funciones):
            if metodo == 'Simpson':
                # Asegurar que n sea par para Simpson
                n_ajustado = n if n % 2 == 0 else n - 1
                if n_ajustado < 2:  # Mínimo 2 subintervalos para Simpson
                    continue
                valor = funcion(a, b, n_ajustado)
                nodos_simpson = n_ajustado + 1
                resultados[metodo]['nodos'].append(nodos_simpson)
            else:
                valor = funcion(a, b, n)
                resultados[metodo]['nodos'].append(nodos)
            
            resultados[metodo]['valores'].append(valor)
            resultados[metodo]['errores'].append(abs(valor - exacto))
    
    # Encontrar número de nodos necesario para cada método
    print("\n" + "="*60)
    print("NODOS NECESARIOS PARA PRECISIÓN DE 1e-7")
    print("="*60)
    
    for metodo in metodos:
        errores = resultados[metodo]['errores']
        nodos_lista = resultados[metodo]['nodos']
        
        # Buscar primer error menor que la precisión
        encontrado = False
        for i, error in enumerate(errores):
            if error < precision:
                print(f"{metodo:12s}: {nodos_lista[i]:4d} nodos (error = {error:.2e})")
                encontrado = True
                break
        
        if not encontrado and errores:
            min_error = min(errores)
            min_nodos = nodos_lista[errores.index(min_error)]
            print(f"{metodo:12s}: > {min_nodos:4d} nodos (error mínimo = {min_error:.2e})")
    
    return resultados, metodos


def graficar_convergencia(resultados, metodos):
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    fig.suptitle('Convergencia de Métodos de Integración Numérica', fontsize=16)
    
    exacto = valor_exacto()
    precision = 1e-7
    
    # Colores para cada método
    colores = {'Riemann': 'blue', 'Punto Medio': 'green', 
               'Trapezoide': 'red', 'Simpson': 'orange'}
    
    # Gráfico 1: Valores de la integral vs nodos
    ax1 = axes[0, 0]
    for metodo in metodos:
        ax1.plot(resultados[metodo]['nodos'], resultados[metodo]['valores'], 
                color=colores[metodo], label=metodo, linewidth=1.5, alpha=0.8)
    ax1.axhline(y=exacto, color='black', linestyle='--', linewidth=1, label='Valor Exacto')
    ax1.set_xlabel('Número de Nodos')
    ax1.set_ylabel('Valor de la Integral')
    ax1.set_title('Convergencia de los Métodos')
    ax1.grid(True, alpha=0.3)
    ax1.legend()
    ax1.set_ylim(1.26424, 1.264242)  # Zoom para ver diferencias
    
    # Gráfico 2: Error absoluto vs nodos (escala logarítmica)
    ax2 = axes[0, 1]
    for metodo in metodos:
        if resultados[metodo]['errores']:
            ax2.semilogy(resultados[metodo]['nodos'], resultados[metodo]['errores'], 
                        color=colores[metodo], label=metodo, linewidth=1.5, alpha=0.8)
    ax2.axhline(y=precision, color='purple', linestyle='--', linewidth=1, 
               label=f'Precisión 1e-7')
    ax2.set_xlabel('Número de Nodos')
    ax2.set_ylabel('Error Absoluto (log)')
    ax2.set_title('Error Absoluto vs Nodos')
    ax2.grid(True, alpha=0.3, which='both')
    ax2.legend()
    
    # Gráfico 3: Error relativo vs nodos
    ax3 = axes[1, 0]
    for metodo in metodos:
        if resultados[metodo]['valores']:
            errores_relativos = [abs(v - exacto)/abs(exacto) for v in resultados[metodo]['valores']]
            ax3.semilogy(resultados[metodo]['nodos'], errores_relativos, 
                        color=colores[metodo], label=metodo, linewidth=1.5, alpha=0.8)
    ax3.axhline(y=precision/abs(exacto), color='purple', linestyle='--', linewidth=1)
    ax3.set_xlabel('Número de Nodos')
    ax3.set_ylabel('Error Relativo (log)')
    ax3.set_title('Error Relativo vs Nodos')
    ax3.grid(True, alpha=0.3, which='both')
    ax3.legend()
    
    # Gráfico 4: Comparación directa de convergencia
    ax4 = axes[1, 1]
    
    max_nodos_visual = 100
    for metodo in metodos:
        nodos_filtrados = []
        valores_filtrados = []
        for n, v in zip(resultados[metodo]['nodos'], resultados[metodo]['valores']):
            if n <= max_nodos_visual:
                nodos_filtrados.append(n)
                valores_filtrados.append(v)
        if nodos_filtrados:
            ax4.plot(nodos_filtrados, valores_filtrados, 
                    color=colores[metodo], label=metodo, linewidth=2, alpha=0.8)
    
    # Líneas horizontales para valor exacto y banda de precisión
    ax4.axhline(y=exacto, color='black', linestyle='-', linewidth=1.5, label='Exacto')
    ax4.axhline(y=exacto + precision, color='gray', linestyle=':', linewidth=1, alpha=0.5)
    ax4.axhline(y=exacto - precision, color='gray', linestyle=':', linewidth=1, alpha=0.5)
    ax4.fill_between([0, max_nodos_visual], exacto - precision, exacto + precision, 
                    alpha=0.1, color='gray', label='Banda de precisión 1e-7')
    
    ax4.set_xlabel('Número de Nodos')
    ax4.set_ylabel('Valor de la Integral')
    ax4.set_title(f'Zoom: Primeros {max_nodos_visual} nodos')
    ax4.grid(True, alpha=0.3)
    ax4.legend()
    ax4.set_xlim(0, max_nodos_visual)
    ax4.set_ylim(1.264240, 1.264242)
    
    plt.tight_layout()
    plt.show()
    
    # Gráfico adicional: Comparación de eficiencia
    plt.figure(figsize=(10, 6))
    for metodo in metodos:
        if resultados[metodo]['errores']:
            plt.loglog(resultados[metodo]['nodos'], resultados[metodo]['errores'], 
                      color=colores[metodo], label=metodo, linewidth=2, alpha=0.8)
    
    plt.axhline(y=precision, color='purple', linestyle='--', linewidth=1.5, 
               label=f'Precisión 1e-7')
    plt.xlabel('Número de Nodos (escala log)')
    plt.ylabel('Error Absoluto (escala log)')
    plt.title('Comparación de Eficiencia: Tasa de Convergencia')
    plt.grid(True, alpha=0.3, which='both')
    plt.legend()
    plt.show()

# Ejecutar el estudio de convergencia
print("ESTUDIO DE CONVERGENCIA - PRECISIÓN DE 7 CIFRAS DECIMALES")
print("="*60)
resultados, metodos = estudio_convergencia()

# Generar gráficos
graficar_convergencia(resultados, metodos)

```

- El error local en el subintervalo que contiene x=0 es pequeño porque $$f(0)=0$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from scipy.special import erf


def f(x, y):
    """y' = x*y + x^2"""
    return x * y + x**2

# ==================== SOLUCIÓN EXACTA ANALÍTICA ====================
def exact_solution_analytic(x):
   
    # Para evitar problemas numéricos con x=0
    x = np.asarray(x, dtype=float)
    result = np.zeros_like(x)
    
    mask = (x == 0)
    result[mask] = 1.0
    
    mask_nonzero = (x != 0)
    x_nonzero = x[mask_nonzero]
    result[mask_nonzero] = -x_nonzero + np.exp(x_nonzero**2/2) * (
        1 + np.sqrt(np.pi/2) * erf(x_nonzero/np.sqrt(2))
    )
    
    return result

# ==================== MÉTODO DE HEUN ====================
def heun(f, x0, y0, xf, n):
    
    h = (xf - x0) / n
    x = np.linspace(x0, xf, n + 1)
    y = np.zeros(n + 1)
    y[0] = y0
    
    for i in range(n):
        k1 = f(x[i], y[i])
        k2 = f(x[i] + h, y[i] + h * k1)
        y[i+1] = y[i] + (h / 2) * (k1 + k2)
    
    return x, y

# ==================== GRÁFICO PRINCIPAL ====================
def grafico_comparativo():
    # Configurar figura
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    
    # 1. Gráfico principal: solución exacta vs Heun
    ax1 = axes[0, 0]
    
    # Solución exacta ANALÍTICA (no numérica)
    x_exact = np.linspace(0, 1, 1000)
    y_exact = exact_solution_analytic(x_exact)
    ax1.plot(x_exact, y_exact, 'k-', linewidth=3, label='Solución Exacta Analítica', alpha=0.8)
    
    # Colores y marcadores para diferentes números de nodos
    colors = ['red', 'blue', 'green', 'purple']
    markers = ['o', 's', '^', 'd']
    nodos = [50, 100, 150, 200]
    
    for idx, n in enumerate(nodos):
        x_heun, y_heun = heun(f, 0, 1, 1, n)
        
        # Calcular error máximo
        y_exact_at_nodes = exact_solution_analytic(x_heun)
        error_max = np.max(np.abs(y_heun - y_exact_at_nodes))
        
        ax1.plot(x_heun, y_heun, color=colors[idx], marker=markers[idx], 
                markersize=4, linewidth=1.5, linestyle='--',
                label=f'Heun (n={n}), Error max={error_max:.2e}')
    
    ax1.set_xlabel('x', fontsize=12)
    ax1.set_ylabel('y(x)', fontsize=12)
    ax1.set_title('Método de Heun vs Solución Exacta Analítica', fontsize=14)
    ax1.grid(True, alpha=0.3)
    ax1.legend()
    ax1.set_ylim(0.8, 3.2)
    
    # 2. Gráfico de errores absolutos
    ax2 = axes[0, 1]
    
    for idx, n in enumerate(nodos):
        x_heun, y_heun = heun(f, 0, 1, 1, n)
        y_exact_at_nodes = exact_solution_analytic(x_heun)
        error = np.abs(y_heun - y_exact_at_nodes)
        
        ax2.plot(x_heun, error, color=colors[idx], marker=markers[idx], 
                markersize=3, linewidth=1, label=f'n={n}')
    
    ax2.set_xlabel('x', fontsize=12)
    ax2.set_ylabel('Error Absoluto', fontsize=12)
    ax2.set_title('Error del Método de Heun', fontsize=14)
    ax2.grid(True, alpha=0.3)
    ax2.legend()
    
    # 3. Gráfico de convergencia
    ax3 = axes[1, 0]
    
    nodos_test = np.arange(10, 301, 10)
    errors_max = []
    
    for n in nodos_test:
        x_heun, y_heun = heun(f, 0, 1, 1, n)
        y_exact_at_nodes = exact_solution_analytic(x_heun)
        error_max = np.max(np.abs(y_heun - y_exact_at_nodes))
        errors_max.append(error_max)
    
    h_values = 1 / nodos_test
    ax3.loglog(nodos_test, errors_max, 'bo-', linewidth=2, label='Error máximo')
    ax3.loglog(nodos_test, 10 * h_values**2, 'r--', linewidth=2, label='O(h²)')
    
    ax3.set_xlabel('Número de Nodos', fontsize=12)
    ax3.set_ylabel('Error Máximo (log)', fontsize=12)
    ax3.set_title('Convergencia: Error máximo vs Nodos', fontsize=14)
    ax3.grid(True, alpha=0.3, which='both')
    ax3.legend()
    
    # 4. Tabla de resultados en el último subplot
    ax4 = axes[1, 1]
    ax4.axis('off')  # Desactivar ejes para la tabla
    
    # Crear tabla de resultados
    table_data = []
    headers = ['Nodos', 'y(1) aprox', 'y(1) exacto', 'Error abs', 'Error rel %']
    
    for n in nodos:
        x_heun, y_heun = heun(f, 0, 1, 1, n)
        y_exact_1 = exact_solution_analytic([1.0])[0]
        y_approx_1 = y_heun[-1]
        
        error_abs = abs(y_approx_1 - y_exact_1)
        error_rel = 100 * error_abs / abs(y_exact_1)
        
        table_data.append([
            f'{n}',
            f'{y_approx_1:.8f}',
            f'{y_exact_1:.8f}',
            f'{error_abs:.2e}',
            f'{error_rel:.4f}'
        ])
    
    # Crear tabla
    table = ax4.table(
        cellText=table_data,
        colLabels=headers,
        cellLoc='center',
        loc='center',
        colWidths=[0.15, 0.25, 0.25, 0.2, 0.15]
    )
    
    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1.2, 1.8)
    
    ax4.set_title('Resultados en x=1', fontsize=14, y=0.95)
    
    plt.tight_layout()
    plt.show()
    
    # ==================== ANÁLISIS NUMÉRICO ====================
    print("="*70)
    print("ANÁLISIS NUMÉRICO DEL MÉTODO DE HEUN")
    print("="*70)
    print(f"EDO: y' = x*y + x²,  y(0) = 1,  x ∈ [0, 1]")
    print(f"Solución exacta en x=1: y(1) = {exact_solution_analytic([1.0])[0]:.10f}")
    print()
    
    print(f"{'Nodos':<10} {'h':<12} {'y(1) approx':<15} {'Error absoluto':<15} {'Error relativo %':<15}")
    print("-"*70)
    
    for n in nodos:
        x_heun, y_heun = heun(f, 0, 1, 1, n)
        y_exact_1 = exact_solution_analytic([1.0])[0]
        y_approx_1 = y_heun[-1]
        
        error_abs = abs(y_approx_1 - y_exact_1)
        error_rel = 100 * error_abs / abs(y_exact_1)
        
        print(f"{n:<10} {1/n:<12.6f} {y_approx_1:<15.10f} {error_abs:<15.2e} {error_rel:<15.4f}")
    
    print("-"*70)
    
    # Calcular tasa de convergencia
    print("\nTasa de convergencia estimada (p en error ≈ C*h^p):")
    for i in range(len(nodos)-1):
        n1, n2 = nodos[i], nodos[i+1]
        
        x1, y1 = heun(f, 0, 1, 1, n1)
        x2, y2 = heun(f, 0, 1, 1, n2)
        
        y_exact_1 = exact_solution_analytic(x1)
        y_exact_2 = exact_solution_analytic(x2)
        
        error1 = np.max(np.abs(y1 - y_exact_1))
        error2 = np.max(np.abs(y2 - y_exact_2))
        
        h1, h2 = 1/n1, 1/n2
        p = np.log(error1/error2) / np.log(h1/h2)
        
        print(f"Comparando n={n1} y n={n2}: p ≈ {p:.4f} (teórico: p=2)")

# ==================== EJECUCIÓN ====================
if __name__ == "__main__":
    grafico_comparativo()
```
