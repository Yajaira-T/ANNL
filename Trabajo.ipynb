### a) Justifique que $ f \in C^\infty([0,2]) $

Las funciones

pertenecen a $ C^\infty(\mathbb{R}) $ (de hecho, son funciones analíticas).  
Como el producto de funciones $ C^\infty $ es nuevamente $ C^\infty $, se concluye que

$$
f(x)=e^{-x}\cos(5x) \in C^\infty(\mathbb{R}).
$$

En particular, su restricción al intervalo cerrado $ [0,2] $ pertenece a

### b) Explique por qué la integral $I_f = \int_0^2 f(x)\,dx$ existe y es única

Dado que  

$$
f \in C^\infty([0,2]),
$$

en particular $f$ es continua en el intervalo cerrado y acotado $[0,2]$.

Por el **teorema fundamental de la integrabilidad de Riemann**, toda función continua en un intervalo cerrado y acotado es integrable. Por tanto, la integral

$$
I_f=\int_0^2 f(x)\,dx
$$

**existe**.

La **unicidad** se garantiza porque la integral definida de Riemann (y también la de Lebesgue) está bien definida: si una función es integrable, el valor del integral es único y no depende del método de partición.

Además, como $f$ es continua, existe una primitiva $F$ y, por el **Teorema Fundamental del Cálculo**,

$$
\int_0^2 f(x)\,dx = F(2)-F(0),
$$

lo cual fija nuevamente un valor único.

## Ejercicio 3: Aproximación de la integral con reglas clásicas

Se desea aproximar

$$
I_f=\int_{0}^{2} f(x)\,dx,
\qquad
f(x)=e^{-x}\cos(5x),
$$

mediante reglas compuestas en una partición uniforme.

Sea un número de subintervalos

$$
n\in\mathbb{N},
$$

y defínase el paso

$$
h=\frac{b-a}{n}=\frac{2-0}{n}=\frac{2}{n},
$$

con nodos

$$
x_i=a+ih,\qquad i=0,1,\dots,n.
$$



### a) Regla compuesta del punto medio

En cada subintervalo

$$
[x_{i-1},x_i],
$$

el punto medio es

$$
m_i=\frac{x_{i-1}+x_i}{2}.
$$

La regla compuesta del punto medio es

$$
M_n
=
h\sum_{i=1}^{n} f(m_i).
$$



### b) Regla compuesta del trapecio

La regla compuesta del trapecio es

$$
T_n
=
h\left[\frac{f(x_0)+f(x_n)}{2}+\sum_{i=1}^{n-1} f(x_i)\right].
$$



### c) Regla compuesta de Simpson

La regla compuesta de Simpson requiere

$$
n \text{ par}.
$$

La aproximación es

$$
S_n
=
\frac{h}{3}
\left[
f(x_0)+f(x_n)
+4\sum_{i=1,\, i\ \text{impar}}^{n-1} f(x_i)
+2\sum_{i=2,\, i\ \text{par}}^{n-2} f(x_i)
\right].
$$


```python
# Ejercicio 3: Implementación de las reglas compuestas 

import mpmath as mp

mp.mp.dps = 80  # precisión suficiente para cálculos numéricos 

def funcion_f(x):
    return mp.e**(-x) * mp.cos(5*x)

def regla_punto_medio_compuesta(funcion, a, b, n):
    n = int(n)
    h = (b - a) / n
    suma = mp.mpf('0')
    for i in range(1, n + 1):
        x_izq = a + (i - 1) * h
        x_der = a + i * h
        punto_medio = (x_izq + x_der) / 2
        suma += funcion(punto_medio)
    return h * suma

def regla_trapecio_compuesta(funcion, a, b, n):
    n = int(n)
    h = (b - a) / n
    suma = (funcion(a) + funcion(b)) / 2
    for i in range(1, n):
        suma += funcion(a + i * h)
    return h * suma

def regla_simpson_compuesta(funcion, a, b, n):
    n = int(n)
    if n % 2 != 0:
        raise ValueError("La regla de Simpson compuesta requiere que n sea par.")
    h = (b - a) / n

    suma_extremos = funcion(a) + funcion(b)

    suma_impares = mp.mpf('0')
    for i in range(1, n, 2):
        suma_impares += funcion(a + i * h)

    suma_pares = mp.mpf('0')
    for i in range(2, n, 2):
        suma_pares += funcion(a + i * h)

    return (h / 3) * (suma_extremos + 4 * suma_impares + 2 * suma_pares)

# Ejemplo de ejecución 
a = mp.mpf('0')
b = mp.mpf('2')
n = 4

aprox_punto_medio = regla_punto_medio_compuesta(funcion_f, a, b, n)
aprox_trapecio = regla_trapecio_compuesta(funcion_f, a, b, n)
aprox_simpson = regla_simpson_compuesta(funcion_f, a, b, n)

print(f"Aproximación de I_f por regla del punto medio compuesta con n = {n}:")
print(aprox_punto_medio)

print(f"\nAproximación de I_f por regla del trapecio compuesta con n = {n}:")
print(aprox_trapecio)

print(f"\nAproximación de I_f por regla de Simpson compuesta con n = {n}:")
print(aprox_simpson)
```

## Ejercicio 5: Error absoluto y orden empírico de convergencia

Se considera como **valor verdadero** (referencia) el valor de alta precisión

$$
I_f=\int_{0}^{2} e^{-x}\cos(5x)\,dx.
$$

Para cada método (punto medio, trapecio y Simpson) y para

$$
n=4 \quad \text{y} \quad n=8,
$$

se calcula el **error absoluto**:

$$
E_n=\left|I_f - A_n\right|,
$$

donde $$A_n$$ es la aproximación numérica correspondiente.



### a) Error absoluto en cada aproximación

Se reportan los errores $$E_4$$ y $$E_8$$ para cada método.


### b) Orden empírico al duplicar el número de subintervalos

Al duplicar $$n$$ (de 4 a 8), el **orden empírico** se estima mediante:

$$
p_{\text{emp}}
=
\frac{\ln\left(E_4/E_8\right)}{\ln(2)}.
$$



### c) Comentario sobre el comportamiento de Simpson

Como $$f(x)=e^{-x}\cos(5x)$$ es una función suave (de clase $$C^\infty$$ en $$[0,2]$$), se espera teóricamente:

$$
E_n=\mathcal{O}(h^2) \quad \text{para punto medio y trapecio},
\qquad
E_n=\mathcal{O}(h^4) \quad \text{para Simpson}.
$$

Por ello, al refinar la malla, la regla de Simpson típicamente reduce el error mucho más rápido.
En presencia de oscilaciones (por el término $$\cos(5x)$$), el error puede mostrar cancelaciones parciales o variaciones en la constante del error, pero la ventaja del orden $$4$$ de Simpson suele mantenerse para tamaños de paso suficientemente pequeños.


```python
pip install pandas 
```

```python


import mpmath as mp
import pandas as pd

mp.mp.dps = 80  # alta precisión

# Intervalo
a = mp.mpf("0")
b = mp.mpf("2")

# Valor verdadero (referencia)
I_verdadero = (mp.e**(-2) * (-mp.cos(10) + 5*mp.sin(10)) + 1) / 26

print("Valor verdadero de referencia I_f =")
print(I_verdadero)

# Aproximaciones para n = 4 y n = 8
aproximaciones = {
    4: {
        "Punto medio": regla_punto_medio_compuesta(funcion_f, a, b, 4),
        "Trapecio": regla_trapecio_compuesta(funcion_f, a, b, 4),
        "Simpson": regla_simpson_compuesta(funcion_f, a, b, 4)
    },
    8: {
        "Punto medio": regla_punto_medio_compuesta(funcion_f, a, b, 8),
        "Trapecio": regla_trapecio_compuesta(funcion_f, a, b, 8),
        "Simpson": regla_simpson_compuesta(funcion_f, a, b, 8)
    }
}

# Errores absolutos
errores = {}
for n in [4, 8]:
    errores[n] = {}
    for metodo in aproximaciones[n]:
        errores[n][metodo] = mp.fabs(I_verdadero - aproximaciones[n][metodo])

print("\nErrores absolutos para n = 4:")
for metodo in errores[4]:
    print(f"Error absoluto con {metodo} (n=4) =")
    print(errores[4][metodo])

print("\nErrores absolutos para n = 8:")
for metodo in errores[8]:
    print(f"Error absoluto con {metodo} (n=8) =")
    print(errores[8][metodo])

# Orden empírico
ordenes = {}
for metodo in errores[4]:
    ordenes[metodo] = mp.log(errores[4][metodo] / errores[8][metodo]) / mp.log(2)

print("\nOrden empírico de convergencia (n = 4 → 8):")
for metodo in ordenes:
    print(f"Orden empírico estimado para {metodo} =")
    print(ordenes[metodo])

# Tabla resumen
tabla_ej5 = pd.DataFrame([
    {
        "Método": metodo,
        "Error absoluto n=4": errores[4][metodo],
        "Error absoluto n=8": errores[8][metodo],
        "Orden empírico": ordenes[metodo]
    }
    for metodo in errores[4]
])

print("\nTabla resumen del Ejercicio 5:")
tabla_ej5
```

## Ejercicio 7: Comparación de eficiencia y efecto de la naturaleza oscilatoria

Se comparan los métodos aplicados en el Ejercicio 6 para la integral

$$
I_f=\int_{0}^{2} e^{-x}\cos(5x)\,dx,
$$

utilizando una partición uniforme con

$$
n=16
$$

subintervalos.


### Comparación de la eficiencia de los métodos

La **eficiencia** de un método de integración numérica se evalúa considerando:

- la **precisión alcanzada** (error absoluto),
- el **orden de convergencia**,
- y el **costo computacional**, medido principalmente por el número de evaluaciones de la función.

La regla compuesta del trapecio presenta un **orden teórico de convergencia 2**, mientras que la regla compuesta de Simpson presenta un **orden teórico 4**.  
Esto implica que, para un mismo tamaño de paso $$h$$, Simpson reduce el error de forma mucho más rápida.

Para una partición con $$n=16$$:

- La regla del trapecio requiere $$n+1=17$$ evaluaciones de la función.
- La regla de Simpson también utiliza los mismos nodos, pero combina la información con ponderaciones que permiten capturar mejor la curvatura local de la función.

Por tanto, **a igual número de subintervalos**, la regla de Simpson resulta claramente **más eficiente**, ya que alcanza una precisión significativamente mayor sin incrementar sustancialmente el costo computacional.



### Efecto de la naturaleza oscilatoria de $$\cos(5x)$$ en el error

La función integrando

$$
f(x)=e^{-x}\cos(5x)
$$

presenta un comportamiento **oscilatorio** debido al término $$\cos(5x)$$, con frecuencia relativamente alta en el intervalo $$[0,2]$$. Esta oscilación tiene varios efectos sobre los métodos numéricos:

1. **Cancelaciones parciales del error**  
   Las oscilaciones pueden provocar que errores locales positivos y negativos se compensen parcialmente, especialmente en métodos de bajo orden como el trapecio.

2. **Sensibilidad al tamaño de paso**  
   Si el paso $$h$$ no es suficientemente pequeño para resolver adecuadamente las oscilaciones, el método puede no capturar correctamente la forma de la función, incrementando el error.

3. **Ventaja de Simpson en funciones suaves**  
   Dado que $$f$$ es una función suave ($$C^\infty$$), la regla de Simpson aprovecha la información adicional sobre la curvatura mediante interpolación cuadrática, lo que le permite manejar mejor las oscilaciones cuando la malla es suficientemente fina.

En consecuencia, aunque ambos métodos se ven afectados por la naturaleza oscilatoria de la función, **Simpson mantiene un mejor comportamiento del error** para valores moderados de $$n$$, mientras que el trapecio requiere un refinamiento mayor de la malla para alcanzar una precisión comparable.



**Conclusión:**  
Para funciones suaves y oscilatorias como $$e^{-x}\cos(5x)$$, la regla compuesta de Simpson es más eficiente que la regla del trapecio, ya que logra una mayor precisión con un costo computacional similar.

## Ejercicio 8: Tabla de Romberg basada en la regla del trapecio

Se construye la tabla de Romberg para

$$
I_f=\int_{0}^{2} e^{-x}\cos(5x)\,dx,
\qquad
f(x)=e^{-x}\cos(5x),
$$

usando la regla del trapecio con los pasos

$$
h,\quad \frac{h}{2},\quad \frac{h}{4}.
$$

En Romberg, el paso inicial es

$$
h=b-a=2,
$$

por lo que los refinamientos corresponden a:

$$
h=2 \Longleftrightarrow n=1,\qquad
\frac{h}{2}=1 \Longleftrightarrow n=2,\qquad
\frac{h}{4}=0.5 \Longleftrightarrow n=4.
$$

Se define:

$$
R_{0,0}=T(h),\qquad
R_{1,0}=T\left(\frac{h}{2}\right),\qquad
R_{2,0}=T\left(\frac{h}{4}\right),
$$

donde $$T(\cdot)$$ es el trapecio compuesto.

La extrapolación de Richardson en Romberg se construye con:

$$
R_{k,j}
=
R_{k,j-1}
+
\frac{R_{k,j-1}-R_{k-1,j-1}}{4^{j}-1},
\qquad
j\ge 1.
$$

En particular,

$$
R_{1,1}=R_{1,0}+\frac{R_{1,0}-R_{0,0}}{3},
\qquad
R_{2,1}=R_{2,0}+\frac{R_{2,0}-R_{1,0}}{3},
$$

y

$$
R_{2,2}=R_{2,1}+\frac{R_{2,1}-R_{1,1}}{15}.
$$

El valor solicitado en el literal (b) es $$R_{2,2}$$.


```python


import mpmath as mp
import pandas as pd

mp.mp.dps = 80

# Intervalo
a = mp.mpf("0")
b = mp.mpf("2")

# Trapecio con n = 1, 2, 4 (equivalente a pasos h, h/2, h/4 con h=b-a)
T1 = regla_trapecio_compuesta(funcion_f, a, b, 1)  # h
T2 = regla_trapecio_compuesta(funcion_f, a, b, 2)  # h/2
T4 = regla_trapecio_compuesta(funcion_f, a, b, 4)  # h/4

print("Aproximación por trapecio compuesto con n = 1 (paso h = 2) =")
print(T1)

print("\nAproximación por trapecio compuesto con n = 2 (paso h/2 = 1) =")
print(T2)

print("\nAproximación por trapecio compuesto con n = 4 (paso h/4 = 0.5) =")
print(T4)

# Construcción de la tabla de Romberg
R00 = T1
R10 = T2
R20 = T4

R11 = R10 + (R10 - R00) / (4**1 - 1)   # /3
R21 = R20 + (R20 - R10) / (4**1 - 1)   # /3

R22 = R21 + (R21 - R11) / (4**2 - 1)   # /15

print("\nAproximación Romberg R_{1,1} =")
print(R11)

print("\nAproximación Romberg R_{2,1} =")
print(R21)

print("\nAproximación solicitada Romberg R_{2,2} =")
print(R22)

# Tabla en formato DataFrame
tabla_romberg = pd.DataFrame(
    [
        [R00, None, None],
        [R10, R11, None],
        [R20, R21, R22]
    ],
    columns=["j=0", "j=1", "j=2"],
    index=["k=0", "k=1", "k=2"]
)

print("\nTabla de Romberg (Ejercicio 8):")
tabla_romberg
```

## Ejercicio 9

Se considera la integral

$$
I_f=\int_0^2 e^{-x}\cos(5x)\,dx,
\qquad
f(x)=e^{-x}\cos(5x).
$$



### 9.a) Hipótesis de suavidad que justifican la extrapolación de Richardson

La extrapolación de Richardson se justifica cuando el error del método base  admite una **expansión asintótica en potencias pares de** $$h$$.

En particular, para el trapecio compuesto se requiere que la función sea suficientemente suave en $$[a,b]$$ para garantizar que exista una expansión del tipo:

$$
T(h) = I_f + c_2 h^2 + c_4 h^4 + c_6 h^6 + \cdots,
$$

donde $$c_2, c_4, c_6,\dots$$ son constantes que dependen de derivadas de $$f$$.

Una condición estándar suficiente es que

$$
f \in C^{2m}([a,b])
$$

para obtener una expansión válida hasta términos de orden $$h^{2m}$$.  
En particular:

- Para eliminar el término principal $$c_2h^2$$ (primera extrapolación, como en $$R_{k,1}$$), basta con que
  $$
  f \in C^2([a,b]).
  $$

- Para obtener de forma consistente una segunda extrapolación (como en $$R_{2,2}$$), es suficiente asumir que
  $$
  f \in C^4([a,b]).
  $$

En este caso se cumple una condición más fuerte, pues

$$
f(x)=e^{-x}\cos(5x)\in C^\infty([0,2]),
$$

por lo que la extrapolación de Richardson está plenamente justificada y la tabla de Romberg es estable desde el punto de vista teórico.



### 9.b) Comparación del costo computacional: Romberg vs Simpson compuesto

Se mide el costo principalmente por el **número de evaluaciones de la función** $$f$$.

#### Costo de Simpson compuesto

Para Simpson compuesto con $$n$$ subintervalos (con $$n$$ par), se evalúa $$f$$ en los $$n+1$$ nodos:

$$
x_0,x_1,\dots,x_n,
$$

por lo que el costo es aproximadamente:

$$
\text{costo}(S_n)\approx n+1.
$$

Además, Simpson alcanza orden:

$$
E_n=\mathcal{O}(h^4).
$$



#### Costo de Romberg (basado en trapecio)

Romberg construye aproximaciones con trapecio refinando la malla:

$$
n=1,\;2,\;4,\;8,\dots
$$

En cada refinamiento, si se reutilizan evaluaciones previas, solo se calculan los valores en los **nuevos puntos medios**.  
Para construir hasta el nivel con $$n=2^k$$, el total acumulado de evaluaciones es:

$$
1 + 2^k + 1
$$

(en términos prácticos, es del orden de $$2^k$$, es decir, similar al costo del trapecio con la malla más fina).

La ventaja es que Romberg, mediante extrapolación, incrementa el orden y puede lograr precisiones altas con menos niveles, pero:

- el método requiere almacenar una tabla triangular,
- hacer combinaciones adicionales (costo aritmético extra),
- y su desempeño depende de que la expansión del error sea regular.



#### Conclusión práctica

- **Simpson compuesto** es muy eficiente para funciones suaves porque obtiene orden 4 con una implementación simple y costo $$\approx n+1$$ evaluaciones.
- **Romberg** puede ser más eficiente si se busca alta precisión, porque la extrapolación puede elevar rápidamente el orden; sin embargo, su costo crece al construir niveles sucesivos y puede ser menos robusto si la función no cumple suavidad o presenta comportamiento irregular (singularidades, no diferenciabilidad, etc.).

En este caso, como $$f\in C^\infty([0,2])$$, ambos funcionan bien, pero Simpson suele ser preferible por simplicidad y alta eficiencia para órdenes moderados, mientras que Romberg es competitivo cuando se desea aumentar el orden de manera sistemática con refinamientos sucesivos.


## Ejercicio 10: Integración adaptativa (Simpson adaptativo) con tolerancia $$10^{-6}$$




```python


import mpmath as mp

mp.mp.dps = 50  # precisión suficiente para tol=1e-6

def simpson_simple(funcion, a, b, fa, fm, fb):
    h = b - a
    return (h / 6) * (fa + 4*fm + fb)

def simpson_adaptativo(funcion, a, b, tolerancia):
    # Contador de evaluaciones
    evaluaciones = {"n": 0}

    def evaluar(x):
        evaluaciones["n"] += 1
        return funcion(x)

    def recursivo(a, b, fa, fm, fb, S, tol, profundidad):
        m = (a + b) / 2
        lm = (a + m) / 2
        rm = (m + b) / 2

        flm = evaluar(lm)
        frm = evaluar(rm)

        S_izq = simpson_simple(funcion, a, m, fa, flm, fm)
        S_der = simpson_simple(funcion, m, b, fm, frm, fb)

        # Estimación clásica del error en Simpson adaptativo
        error_estimado = mp.fabs(S_izq + S_der - S) / 15

        # Criterio de parada
        if error_estimado <= tol or profundidad <= 0:
            # Corrección de Richardson para mejorar la aproximación
            return (S_izq + S_der) + (S_izq + S_der - S) / 15

        # Refinamiento adaptativo
        return (recursivo(a, m, fa, flm, fm, S_izq, tol/2, profundidad - 1) +
                recursivo(m, b, fm, frm, fb, S_der, tol/2, profundidad - 1))

    # Valores iniciales
    m = (a + b) / 2
    fa = evaluar(a)
    fm = evaluar(m)
    fb = evaluar(b)

    S0 = simpson_simple(funcion, a, b, fa, fm, fb)

   
    profundidad_max = 50

    aproximacion = recursivo(a, b, fa, fm, fb, S0, mp.mpf(tolerancia), profundidad_max)

    return aproximacion, evaluaciones["n"]

# Parámetros del problema
a = mp.mpf("0")
b = mp.mpf("2")
tolerancia = mp.mpf("1e-6")

If_adaptativo, numero_evaluaciones = simpson_adaptativo(funcion_f, a, b, tolerancia)

print("Aproximación de I_f mediante Simpson adaptativo con tolerancia 1e-6 =")
print(If_adaptativo)

print("\nNúmero de evaluaciones de la función en Simpson adaptativo =")
print(numero_evaluaciones)
```

### 11.b) Distribución de los refinamientos del mallado

El refinamiento del mallado en Simpson adaptativo **no es uniforme**. El método actúa de la siguiente manera:

1. Se comienza con una aproximación de Simpson en todo el intervalo $$[0,2]$$.
2. Se estima el error local comparando la aproximación de Simpson en $$[a,b]$$ con la suma de las aproximaciones en los subintervalos $$[a,m]$$ y $$[m,b]$$.
3. Si el error estimado excede la tolerancia local, el intervalo se subdivide y el proceso se repite recursivamente.

En este problema, la función

$$
f(x)=e^{-x}\cos(5x)
$$

presenta las siguientes características relevantes:

- El término $$\cos(5x)$$ introduce **oscilaciones** relativamente rápidas.
- El factor $$e^{-x}$$ atenúa la amplitud de la función a medida que $$x$$ aumenta.

Como consecuencia:

- El refinamiento del mallado se **concentra principalmente en las zonas donde la función oscila más rápidamente**, es decir, donde el término $$\cos(5x)$$ cambia de signo con mayor frecuencia.
- En regiones donde la función es más suave o tiene menor amplitud (valores grandes de $$x$$ debido al factor exponencial), el método acepta intervalos más grandes sin necesidad de refinamiento adicional.



### Conclusión

- El número de evaluaciones de la función es **adaptativo** y depende directamente de la tolerancia exigida.
- El mallado final es **más fino en regiones problemáticas** (oscilatorias) y **más grueso en regiones suaves**, lo que explica la alta eficiencia del método adaptativo frente a métodos con partición uniforme.
- Este comportamiento automático es una de las principales ventajas de los métodos de integración adaptativa.



## Ejercicio 12: Detección automática de regiones problemáticas y refinamiento del mallado

Se analiza el comportamiento del **método de Simpson adaptativo** aplicado a la integral

$$
I_f=\int_{0}^{2} e^{-x}\cos(5x)\,dx.
$$



### ¿Por qué el método adaptativo detecta automáticamente regiones problemáticas?

El método de Simpson adaptativo se basa en una **estimación local del error**.  
En cada subintervalo $$[a,b]$$, el algoritmo:

1. Calcula una aproximación de Simpson en todo el intervalo $$[a,b]$$.
2. Calcula dos aproximaciones de Simpson en los subintervalos $$[a,m]$$ y $$[m,b]$$, donde
   $$
   m=\frac{a+b}{2}.
   $$
3. Compara ambas aproximaciones y estima el error local mediante:
   $$
   E \approx \frac{|S_{\text{izq}}+S_{\text{der}}-S|}{15}.
   $$

Si el error estimado excede la tolerancia local, el intervalo se subdivide.  
Este mecanismo permite que el método **detecte automáticamente** regiones donde:

- la curvatura es elevada,
- la función oscila rápidamente,
- o las derivadas de orden superior son grandes.

En estas zonas, la aproximación con polinomios de bajo grado (como los de Simpson) es menos precisa, activando el refinamiento.



### ¿En qué zonas del intervalo se concentra el refinamiento y por qué?

La función integrando es

$$
f(x)=e^{-x}\cos(5x).
$$

Sus propiedades relevantes son:

- El término $$\cos(5x)$$ introduce **oscilaciones periódicas** en todo el intervalo.
- El factor $$e^{-x}$$ hace que la **amplitud de las oscilaciones decrezca** conforme $$x$$ aumenta.

Como consecuencia:

- El refinamiento del mallado se concentra principalmente en la **parte izquierda del intervalo**, es decir, cerca de $$x=0$$.
- En esta región, la función tiene **mayor amplitud y curvatura**, lo que genera errores locales más grandes.
- A medida que $$x$$ se acerca a $$2$$, la amplitud de $$f(x)$$ disminuye, las oscilaciones se atenúan y el método acepta intervalos más grandes sin necesidad de subdividir.



### Conclusión

El método adaptativo detecta automáticamente regiones problemáticas porque basa sus decisiones en una **estimación local del error**, sin requerir información previa sobre la función.  
En este problema, el refinamiento se concentra cerca de $$x=0$$ debido a la combinación de **oscilaciones rápidas** y **mayor amplitud**, lo que justifica plenamente el comportamiento observado del mallado adaptativo.


## Ejercicio 13: Integral $$I_g$$

Se considera la integral

$$
I_g=\int_{0}^{1}\sqrt{x}\,\ln(x+1)\,dx,
$$

con

$$
g(x)=\sqrt{x}\,\ln(x+1),\qquad x\in[0,1].
$$


### 13.a) Justificación de que $$g\notin C^2([0,1])$$

Primero, notemos que $$g$$ está bien definida y es continua en $$[0,1]$$, pues $$\sqrt{x}$$ y $$\ln(1+x)$$ son continuas en $$[0,1]$$ y $$\ln(1+0)=0$$.

Derivando para $$x>0$$:

$$
g'(x)
=
\frac{1}{2\sqrt{x}}\ln(1+x)+\frac{\sqrt{x}}{1+x}.
$$

Además, usando el desarrollo de Taylor:

$$
\ln(1+x)=x+\mathcal{O}(x^2)\quad \text{cuando}\quad x\to 0^+,
$$

se obtiene

$$
\frac{1}{2\sqrt{x}}\ln(1+x)
=
\frac{1}{2\sqrt{x}}\left(x+\mathcal{O}(x^2)\right)
=
\frac{1}{2}\sqrt{x}+\mathcal{O}\left(x^{3/2}\right),
$$

y también

$$
\frac{\sqrt{x}}{1+x}=\sqrt{x}+\mathcal{O}\left(x^{3/2}\right).
$$

Por tanto,

$$
g'(x)=\frac{3}{2}\sqrt{x}+\mathcal{O}\left(x^{3/2}\right)\quad \text{cuando}\quad x\to 0^+,
$$

lo que implica que $$g'$$ se extiende continuamente en $$x=0$$ y, por ende, $$g\in C^1([0,1])$$.

Sin embargo, al derivar una vez más (para $$x>0$$) aparecen términos del tipo $$x^{-3/2}\ln(1+x)$$. Usando nuevamente $$\ln(1+x)\sim x$$ cuando $$x\to 0^+$$, se obtiene el comportamiento asintótico

$$
g''(x)\sim C\,x^{-1/2}\quad \text{cuando}\quad x\to 0^+,
$$

para alguna constante $$C\neq 0$$. En consecuencia,

$$
\lim_{x\to 0^+} g''(x)=\infty,
$$

por lo que $$g''$$ no puede extenderse como función continua en $$[0,1]$$.

Así,

$$
g\notin C^2([0,1]).
$$

### ¿Por qué Simpson pierde eficiencia aquí mientras el método adaptativo mantiene precisión?

La regla de Simpson compuesta obtiene su alta eficiencia (orden teórico $$\mathcal{O}(h^4)$$) cuando la función integrando es suficientemente suave en el intervalo, típicamente requiriendo que derivadas de orden alto (como la cuarta derivada) estén acotadas.

En este caso,

$$
g(x)=\sqrt{x}\ln(1+x)
$$

no pertenece a

$$
C^2([0,1]),
$$

porque cerca de $$x=0$$ aparecen singularidades del tipo $$x^{-1/2}$$ en derivadas. Esto rompe las hipótesis de suavidad que sustentan la cota de error de Simpson: el error ya no decrece como $$h^4$$ y, por tanto, Simpson **pierde eficiencia** (requiere un refinamiento mucho mayor para lograr la misma precisión).

En contraste, Simpson adaptativo estima el error **localmente** y refina el mallado **justo donde el error es grande**. Para esta integral, el comportamiento problemático está cerca de

$$
x=0,
$$

por la presencia de $$\sqrt{x}$$. El método adaptativo concentra subdivisiones en esa zona y mantiene precisión global con menos evaluaciones totales que un refinamiento uniforme comparable.

Así, Simpson compuesto sufre por falta de suavidad en $$x=0$$, mientras que el método adaptativo conserva precisión porque detecta y refina automáticamente la región singular.


## Ejercicio 14: 



### 14.a) ¿Qué método elegiría para cada tipo de función?

#### 1) Funciones suaves

Para funciones suaves en un intervalo cerrado, por ejemplo

$$
f \in C^4([a,b]),
$$

elegiría **Simpson compuesto**, porque:

$$
E_n=\mathcal{O}(h^4),
$$

y por tanto alcanza alta precisión con un número moderado de subintervalos, manteniendo un costo computacional bajo.



#### 2) Funciones altamente oscilatorias

Para funciones altamente oscilatorias, elegiría un **método adaptativo** (por ejemplo, **Simpson adaptativo**) o un método que refine el mallado en zonas donde la oscilación no está bien resuelta, porque:

- las reglas compuestas con malla uniforme pueden requerir un $$n$$ muy grande para capturar la oscilación,
- el refinamiento local mejora la eficiencia al colocar más puntos donde el integrando cambia rápido.

En particular, el adaptativo permite ajustar el tamaño de paso para controlar el error sin necesidad de refinar de forma uniforme todo el intervalo.



#### 3) Funciones con singularidades

Para funciones con singularidades (por ejemplo, derivadas no acotadas o comportamiento tipo $$x^{-\alpha}$$ cerca de un extremo), elegiría **integración adaptativa**, porque:

- las reglas de orden alto como Simpson compuesto pierden su eficiencia si no se cumplen hipótesis de suavidad,
- el método adaptativo detecta automáticamente el aumento del error local y concentra el refinamiento en la región singular.



### 14.b) ¿Existe un método de integración numérica universalmente óptimo?

No existe un método de integración numérica que sea universalmente óptimo.

La razón es que el desempeño de un método depende fuertemente de:

- la suavidad del integrando,
- la presencia de oscilaciones,
- singularidades o no diferenciabilidad,
- el costo de evaluar la función,
- y el nivel de precisión requerido.

Por ejemplo:

- Simpson es excelente para funciones suaves,
- Romberg funciona bien cuando existe una expansión regular del error,
- los métodos adaptativos son más robustos ante singularidades o cambios locales bruscos,

pero ninguno domina en todos los escenarios posibles.

**Conclusión:**
La elección del método debe basarse en las propiedades del integrando y en los requisitos de precisión y costo computacional del problema.
